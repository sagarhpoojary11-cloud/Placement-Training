if else

```python
age=int(input("Enter the age"))
if age>=18:
    print("Allowed for Toxic")
else:
    print("Not Allowed for Toxic")
```

if else else

```python
age=int(input("Enter the age"))
if age<=3:
    print("child")
elif age>=4 and age<=12:
    print("kid")
elif age>=13 and age<=18:
    print("Teenage")
elif age>=19 and age<=60:
    print("Adult")
else:
    print("Thatha")
```
